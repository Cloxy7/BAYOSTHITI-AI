```
B='https://pubchem.ncbi.nlm.nih.gov/rest/pug'
rows=[]
for i,n in enumerate(NAMES):
    r=None
    for t in range(3):
        try:
            r=requests.get(f"{B}/compound/name/{urllib.parse.quote(n)}/property/IsomericSMILES,CanonicalSMILES,SMILES,InChIKey,MolecularFormula,IUPACName/JSON",timeout=30)
            if r.status_code in (200,404): break
        except Exception: pass
        time.sleep(2)
    d={}
    if r is not None and r.status_code==200:
        p=r.json()['PropertyTable']['Properties'][0]; d=p
    rows.append(dict(name=n,cid=d.get('CID',''),smiles=d.get('IsomericSMILES') or d.get('SMILES') or d.get('CanonicalSMILES',''),
                     inchikey=d.get('InChIKey',''),formula=d.get('MolecularFormula',''),iupac=d.get('IUPACName','')))
    if i%25==0: print(i,len(NAMES))
    time.sleep(0.25)
df=pd.DataFrame(rows); df.to_csv('pubchem_new_ions.csv',index=False)
print('found',(df.cid!='').sum(),'of',len(df))
from google.colab import files; files.download('pubchem_new_ions.csv')
```
